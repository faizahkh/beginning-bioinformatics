```python
# Complete Rosalind 9 - Task 23.2

from Bio.SeqIO import parse
from Bio.SeqUtils import gc_fraction

max_id = ''
max_gc = 0

for seq_record in parse("rosalind_gc_1224_1_dataset.txt", "fasta"):
  gc = gc_fraction(seq_record.seq)*100
  if gc > max_gc:
    max_gc = gc
    max_id = seq_record.id

print(max_id)
print(max_gc)
```
